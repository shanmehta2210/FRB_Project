```python
import os
import astrophot as ap
import numpy as np
import torch
from astropy.io import fits
from astropy.wcs import WCS
import matplotlib.pyplot as plt 
from time import time
from matplotlib.ticker import MaxNLocator, FuncFormatter
%matplotlib inline
```

```python
# Build file lists — every cropped galaxy + its matching downsampled PSF
galaxy_dir = "cropped_host_galaxies"
psf_dir = "psfs/downsampled_psfs"

galaxy_files = sorted([f for f in os.listdir(galaxy_dir) if f.endswith("_flux.fits")])
frb_names = [f.replace("_flux.fits", "") for f in galaxy_files]

print(f"Found {len(galaxy_files)} galaxies:")
for name in frb_names:
    gal_path = os.path.join(galaxy_dir, f"{name}_flux.fits")
    psf_path = os.path.join(psf_dir, f"{name}_1x_psf.fits")
    status = "OK" if os.path.exists(psf_path) else "MISSING PSF"
    print(f"  {name}: {status}")
```

```python
# === Run AstroPhot Sersic fitting WITH PSF for every FRB ===
# This is the EXACT same fitting logic as inclination_angle_autoprof.ipynb,
# but with a PSF_Image passed to the model and looped over all galaxies.
# Now uses real variance maps derived from sigma maps (variance = sigma^2).

import csv

results = []

for frb_name in frb_names:
    print(f"\n{'='*60}")
    print(f"Processing {frb_name}")
    print(f"{'='*60}")
    
    # ---- Load the FRB image (same as original notebook) ----
    file_path = os.path.join(galaxy_dir, f"{frb_name}_flux.fits")
    FRB_name = file_path.split("/")[-1]
    
    hdu = fits.open(file_path)
    hdu[0].data = hdu[0].data
    wcs = WCS(hdu[0].header)
    header = hdu[0].header
    target_data = np.array(hdu[0].data, dtype=np.float64)
    
    # ---- Get pixelscale (same as original notebook) ----
    try:
        pixelscale = abs(hdu[0].header['CD2_2']) * 3600
    except KeyError:
        try:
            pixelscale = hdu[0].header['CDELT2'] * 3600
        except KeyError:
            pixelscale = 0.258
    
    # ---- Load sigma map and compute variance (sigma^2) ----
    sigma_path = os.path.join(galaxy_dir, f"{frb_name}_sigma.fits")
    sigma_hdu = fits.open(sigma_path)
    sigma_data = np.array(sigma_hdu[0].data, dtype=np.float64)
    sigma_hdu.close()
    
    # AstroPhot expects variance = sigma^2
    variance_data = sigma_data ** 2
    # Replace any non-finite or zero variance with a large value (effectively low weight)
    bad_mask = ~np.isfinite(variance_data) | (variance_data <= 0)
    if np.any(bad_mask):
        print(f"  Warning: {np.sum(bad_mask)} bad variance pixels replaced")
        variance_data[bad_mask] = np.nanmax(variance_data[np.isfinite(variance_data) & (variance_data > 0)]) * 10
    
    print(f"  Variance map shape: {variance_data.shape}, range: [{variance_data.min():.6g}, {variance_data.max():.6g}]")
    
    # ---- Make a target with real variance ----
    target = ap.image.Target_Image(
        data=target_data,
        pixelscale=pixelscale,
        variance=variance_data,
    )
    
    # ---- Load the downsampled PSF ----
    psf_path = os.path.join(psf_dir, f"{frb_name}_1x_psf.fits")
    psf_hdu = fits.open(psf_path)
    psf_data = np.array(psf_hdu[0].data, dtype=np.float64)
    psf_hdu.close()
    
    psf_image = ap.image.PSF_Image(
        data=psf_data,
        pixelscale=pixelscale,
    )
    
    # ---- Create AstroPhot model WITH PSF (same model_type as original) ----
    model = ap.models.AstroPhot_Model(
        name="model with target",
        model_type="sersic galaxy model",
        target=target,
        psf=psf_image,
    )
    
    model.initialize()
    
    # ---- Fit (same as original notebook) ----
    fitter = ap.fit.LM(model, verbose=1)
    result = fitter.fit()
    print("Fit message:", result.message)
    
    # ---- Extract parameters (same as original notebook) ----
    q = model["q"].value.item()
    inclination_angle = np.degrees(np.arccos(q))
    error_q = model["q"].uncertainty.item()
    error_i = 1 / np.sqrt(1 - q**2 * error_q)
    n_sersic = model["n"].value.item()
    PA = model["PA"].value.item()
    Re = model["Re"].value.item()
    
    # ---- Chi-squared ----
    chi2 = fitter.res_loss()
    ndf = fitter.ndf
    chi2_nu = chi2 / ndf
    
    print(f"  q = {q:.4f} ± {error_q:.4f}")
    print(f"  inclination = {inclination_angle:.2f}° ± {error_i:.2f}°")
    print(f"  n = {n_sersic:.4f}, PA = {PA:.4f}, Re = {Re:.4f}")
    print(f"  chi2 = {chi2:.4f}, ndf = {ndf}, chi2/nu = {chi2_nu:.4f}")
    
    results.append({
        "frb_name": frb_name,
        "q": q,
        "inclination_angle": inclination_angle,
        "error_q": error_q,
        "error_i": error_i,
        "n": n_sersic,
        "PA": PA,
        "Re": Re,
        "chi2": chi2,
        "ndf": ndf,
        "chi2_nu": chi2_nu,
        "fit_message": result.message,
    })
    
    # ---- Plot model + residual (same as original notebook) ----
    fig, ax = plt.subplots(1, 2, figsize=(16, 6))
    ap.plots.model_image(fig, ax[0], model)
    ax[0].set_title(f"{frb_name} — Model (with PSF + variance)")
    ap.plots.residual_image(fig, ax[1], model)
    ax[1].set_title(f"{frb_name} — Residual")
    plt.tight_layout()
    plt.show()
    
    hdu.close()

print(f"\nDone! Processed {len(results)} galaxies.")
```

```python
# Save results to CSV
import pandas as pd

df = pd.DataFrame(results)
output_csv = "astrophot_psf_sigma_inclination_angles.csv"
df.to_csv(output_csv, index=False)
print(f"Saved {len(df)} rows to {output_csv}")
df
```

```python
# === Compare AstroPhot PSF+variance vs GALFIT PSF+sigma ===
import math

galfit_sigma = pd.read_csv("galfit_sigma_metrics_summary.csv")
astrophot_sigma = pd.read_csv("astrophot_psf_sigma_inclination_angles.csv")

# Compute GALFIT inclination angles from b/a using Hubble formula
q0 = 0.2
def hubble_inc(b_a, q0=0.2):
    if b_a <= q0:
        return 90.0
    val = (b_a**2 - q0**2) / (1 - q0**2)
    return math.degrees(math.acos(math.sqrt(max(0, min(1, val)))))

galfit_sigma['inc_psf'] = galfit_sigma['b_a_psf'].apply(lambda x: hubble_inc(float(str(x).replace('*',''))))

# Merge on FRB name
merged = pd.merge(
    astrophot_sigma[['frb_name','q','inclination_angle','n','Re','PA','chi2_nu']],
    galfit_sigma[['FRB','b_a_psf','inc_psf','n_psf','re_psf','pa_psf','chi2nu_psf']],
    left_on='frb_name', right_on='FRB'
)

# Compute differences
merged['delta_inc'] = merged['inclination_angle'] - merged['inc_psf']
merged['delta_q'] = merged['q'] - merged['b_a_psf']
merged['delta_n'] = merged['n'] - merged['n_psf']

comp = merged[['frb_name',
               'inclination_angle','inc_psf','delta_inc',
               'q','b_a_psf','delta_q',
               'n','n_psf','delta_n']].copy()
comp.columns = ['FRB',
                'AstroPhot_inc','GALFIT_inc','delta_inc',
                'AstroPhot_q','GALFIT_q','delta_q',
                'AstroPhot_n','GALFIT_n','delta_n']

print(f"Mean delta_inc = {comp['delta_inc'].mean():.2f} deg  |  Std = {comp['delta_inc'].std():.2f} deg")
print(f"Mean delta_q   = {comp['delta_q'].mean():.4f}       |  Std = {comp['delta_q'].std():.4f}")
print(f"Mean delta_n   = {comp['delta_n'].mean():.4f}       |  Std = {comp['delta_n'].std():.4f}")
comp.round(3)
```

```python
# === CDF Bias Plot: AstroPhot PSF+variance vs GALFIT PSF+sigma ===
import random
from scipy import interpolate
from astropy.io import ascii
import matplotlib.font_manager as font_manager

random.seed(42)

# --- Helper functions ---
def hubble_cosi(b_over_a, q0=0.2):
    val = (b_over_a**2 - q0**2) / (1 - q0**2)
    if val < 0: return 0.0
    if val > 1: return 1.0
    return math.sqrt(val)

# --- Load SDSS background for CDF envelope ---
sdss_data = ascii.read("SDSS_catalog.csv")
b_over_a_sdss = []
for i in range(len(sdss_data)):
    try:
        if sdss_data['petroMag_r'][i] <= 21 and sdss_data['expAB_r'][i] > 0.2:
            b_over_a_sdss.append(hubble_cosi(sdss_data['expAB_r'][i]))
    except:
        pass
print(f"SDSS sample: {len(b_over_a_sdss)} galaxies")

# --- Load FRB sample for matching ---
frb_sample = pd.read_csv('frb_sample.txt', sep='\t')
frb_list = frb_sample['FRB'].tolist()

# --- Compute cos(i) for AstroPhot PSF+variance results ---
astrophot_inc = dict(zip(astrophot_sigma['frb_name'], astrophot_sigma['inclination_angle']))
astrophot_cosi = sorted([np.cos(np.radians(astrophot_inc[f])) for f in frb_list])

# --- Compute cos(i) for GALFIT PSF+sigma results ---
galfit_inc = dict(zip(galfit_sigma['FRB'], galfit_sigma['inc_psf']))
galfit_cosi = sorted([np.cos(np.radians(galfit_inc[f])) for f in frb_list])

n_frb = len(astrophot_cosi)

# --- Generate SDSS CDF envelope (68% CI) ---
total_samples = []
for _ in range(10000):
    new_sample = sorted(random.sample(b_over_a_sdss, n_frb))
    idx_norm = [0] + [j / n_frb for j in range(1, n_frb + 1)] + [1]
    new_sample_ext = [0] + new_sample + [1]
    total_samples.append(interpolate.interp1d(new_sample_ext, idx_norm))

means_cdf = []
sigma_down = []
sigma_up = []
for value in np.linspace(0, 1, 100):
    index_sample = sorted([s(value) for s in total_samples])
    means_cdf.append(np.mean(index_sample))
    sigma_down.append(index_sample[1600])
    sigma_up.append(index_sample[8400])

font_prop = font_manager.FontProperties(family='Arial', style='normal', size=7)

# ============================
# Plot 1: AstroPhot CDF alone
# ============================
fig, ax = plt.subplots(figsize=(7, 5))
x_ap = [0] + astrophot_cosi + [1]
y_ap = [0] + [i / n_frb for i in range(1, n_frb + 1)] + [1]
ax.step(x_ap, y_ap, where='mid', color='red', linewidth=1.5, label="AstroPhot (PSF + variance)")
ax.plot(np.linspace(0, 1, 100), means_cdf, color='black', label='SDSS distribution ($m_r < 21$)')
ax.fill_between(np.linspace(0, 1, 100), sigma_down, sigma_up, color='gray', alpha=0.4,
                label='68% Confidence Interval')
ax.plot((0, 1), (0, 1), color='black', linestyle='--', label="Uniform Distribution")
ax.set_ylabel("Cumulative distribution", fontproperties=font_prop, fontsize=9)
ax.set_xlabel("cos(i)", fontproperties=font_prop, fontsize=9)
ax.set_xlim(0, 1); ax.set_ylim(0, 1)
ax2 = ax.twiny()
ax2.set_xlim(ax.get_xlim())
ax2.set_xticks(np.cos(np.radians([90, 78, 66, 53, 37, 0])))
ax2.set_xticklabels(['90', '78', '66', '53', '37', '0'], fontproperties=font_prop)
ax2.set_xlabel('Inclination angle i (degrees)', fontproperties=font_prop, fontsize=9)
ax.legend(fontsize=8)
ax.set_title("CDF Bias — AstroPhot PSF + Variance Map", fontsize=10)
plt.tight_layout()
os.makedirs("plots/plots_astrophot_psf_sigma", exist_ok=True)
plt.savefig("plots/plots_astrophot_psf_sigma/CDF_bias.pdf", bbox_inches="tight", dpi=300)
plt.show()

# ====================================
# Plot 2: AstroPhot vs GALFIT overlay
# ====================================
fig, ax = plt.subplots(figsize=(7, 5))
x_gf = [0] + galfit_cosi + [1]
y_gf = [0] + [i / n_frb for i in range(1, n_frb + 1)] + [1]
ax.step(x_ap, y_ap, where='mid', color='red', linewidth=1.5, label="AstroPhot (PSF + variance)")
ax.step(x_gf, y_gf, where='mid', color='blue', linewidth=1.5, label="GALFIT (PSF + sigma)")
ax.plot(np.linspace(0, 1, 100), means_cdf, color='black', label='SDSS distribution ($m_r < 21$)')
ax.fill_between(np.linspace(0, 1, 100), sigma_down, sigma_up, color='gray', alpha=0.4,
                label='68% Confidence Interval')
ax.plot((0, 1), (0, 1), color='black', linestyle='--', label="Uniform Distribution")
ax.set_ylabel("Cumulative distribution", fontproperties=font_prop, fontsize=9)
ax.set_xlabel("cos(i)", fontproperties=font_prop, fontsize=9)
ax.set_xlim(0, 1); ax.set_ylim(0, 1)
ax2 = ax.twiny()
ax2.set_xlim(ax.get_xlim())
ax2.set_xticks(np.cos(np.radians([90, 78, 66, 53, 37, 0])))
ax2.set_xticklabels(['90', '78', '66', '53', '37', '0'], fontproperties=font_prop)
ax2.set_xlabel('Inclination angle i (degrees)', fontproperties=font_prop, fontsize=9)
ax.legend(fontsize=8)
ax.set_title("CDF Bias — AstroPhot vs GALFIT (both with noise maps)", fontsize=10)
plt.tight_layout()
plt.savefig("plots/plots_astrophot_psf_sigma/CDF_bias_comparison.pdf", bbox_inches="tight", dpi=300)
plt.show()

print("Plots saved to plots/plots_astrophot_psf_sigma/")
```
